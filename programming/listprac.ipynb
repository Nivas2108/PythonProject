```python
#linear search
n=[2,4,1,9,10]
t=9
for i in range(len(n)):
    if n[i] == t:
        print(i)
```

```python
#Binary Search
def binary_search(n,t):
    low = 0
    high = len(n)-1
    while low <= high:
        mid = (low+high)//2
        if n[mid] == t:
            return mid
        elif mid < t:
            low += 1
        else:
            high -= 1
    return -1
n=[1,3,4,9,12,32,44]
t = 44
print(binary_search(n,t))
```
