```python
### equal row matrix
a=[[1,1,1],[2,2,2],[3,3,3]]
c=0
r=len(a[0])
C=len(a)
for i in range(0,r):
    for j in range(0,C):
        if a[i][0]!=a[i][j]:
            print("not a equal row matrix")
            break
    if c!=0:
        break
else:
    print("equal row matrix")
```

```python
## equal column matrix
a=[[1,2,3],[1,2,3],[1,2,3]]
C=len(a)
r=len(a[0])
c=0
for i in range(0,C):
    for j in range(0,r):
        if a[0][i]!=a[j][i]:
            c+=1
            print("not a equal colo=umn matrix")
            break
    if c!=0:
        break
if c==0:
    print("equal col matrix")
```

```python
#### cut space binary search
def search(l,target):
    s=0
    e=len(l)-1
    while s<=e:
        mid=(s+e)//2
        if l[mid]==target:
            return True
        if l[mid]<=l[e]:
            if target>=l[mid] and target<=l[e]:
                s=mid+1
            else:
                e=mid-1
        else:
            if target>=l[s] and target<=l[mid]:
                e=mid-1
            else:
                s=mid+1
    return False
l=[10,12,15,2,3,6,7,8,9]
target=2
search(l,target)
```

```python
###Recursion###
#####Prime####
def recprime(n,i,count):
    if i>n:
        if count==2:
            print("prime")
        else:
            print("not prime")
        return
    if n%i==0:
        count+=1
    recprime(n,i+1,count)
recprime(7,1,0)
```

```python
###printing numbers in the given range using recursion##
def print_numbers(start, end):
    if start > end:
        return
    print(start,end=" ")
    print_numbers(start + 1, end)
print_numbers(1,10)
```

```python
## printing all digits of a number using recursion##
def print_digits(n):
    if n == 0:
        return
    print_digits(n // 10)
    print(n % 10, end=" ")
print_digits(12345)
```

```python
## divide and conquer merge sort
def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result
def merge_sort(arr):
    if len(arr) == 1:
        return arr
    mid = len(arr) // 2
    left_half = merge_sort(arr[:mid])
    right_half = merge_sort(arr[mid:])
    return merge(left_half, right_half)
l=[23,25,34,35,47,69,20,30,33,42]
sorted_list = merge_sort(l)
print(sorted_list)
```
