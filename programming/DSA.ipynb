```python
### equal row matrix
a=[[1,1,1],[2,2,2],[3,3,3]]
c=0
r=len(a[0])
C=len(a)
for i in range(0,r):
    for j in range(0,C):
        if a[i][0]!=a[i][j]:
            print("not a equal row matrix")
            break
    if c!=0:
        break
else:
    print("equal row matrix")
```

```python
## equal column matrix
a=[[1,2,3],[1,2,3],[1,2,3]]
C=len(a)
r=len(a[0])
c=0
for i in range(0,C):
    for j in range(0,r):
        if a[0][i]!=a[j][i]:
            c+=1
            print("not a equal colo=umn matrix")
            break
    if c!=0:
        break
if c==0:
    print("equal col matrix")
```

```python
### matrix multiplication
A=[[2,3,4],[6,7,5]]
B=[[2,3,4],[3,8,9],[33,7,6]]
c1=len(A[0])
r1=len(A)
r2=len(B)
c2=len(B[0])
for i in range(0,r1):
    for j in range(0,c2):
        sum=0
        for k in range(0,c1):
            sum+=A[i][k]*B[k][j]
        print(sum,end=" ")
    print()
```

```python
#### Merge intervals ####
l=[[15,23],[8,12],[6,7],[1,5],[2,7]]
l=sorted(l,key=lambda x:x[0])
result=[l[0]]
for i in range(1,len(l1)):
    if result[-1][1]>l[i][0]:
        result[-1][1]=max(result[-1][1],l[i][1])
    else:
        result.append(l[i])
print(result)

```

```python
##### Binary Search lower bound
l=[5,6,6,6,9,9,12,12,13,14]
tar=12
s=0
e=len(l)-1
ans=len(l)
while s<=e:
    mid=(s+e)//2
    if l[mid]>=tar:
        ans=mid
        e=mid-1
    else:
        s=mid+1
print(ans)
```

```python
##### Binary Search upper bound
l=[5,6,6,6,9,9,12,12,13,14]
tar=12
s=0
e=len(l)-1
ans=len(l)
while s<=e:
    mid=(s+e)//2
    if l[mid]>tar:
        ans=mid
        e=mid-1
    else:
        s=mid+1
print(ans)
```
